```python
import cv2 as cv
import matplotlib.pylab as plt
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM.jpeg")
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img = cv.resize(img, (0, 0), fx=0.5, fy=0.5)

# POLINEMA
cv.rectangle(img, pt1=(120, 35), pt2=(300, 95), color=(0, 0, 0), thickness=2)
# BNI
cv.rectangle(img, pt1=(565, 40), pt2=(690,95), color=(0, 0, 0), thickness=2)
# PRODI
cv.rectangle(img, pt1=(165, 420), pt2=(430, 390), color=(0, 0, 0), thickness=2)

plt.imshow(img)
plt.show()
```
